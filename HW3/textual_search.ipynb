# Sparse Search - TF-IDF
### Implement

```python
import re

def computeTF(token: str, doc: str):
    # normalize to lowercase
    token = token.lower()
    doc = doc.lower()

    # split document into words
    words = re.findall(r"[a-z]+", doc)

    if len(words) == 0:
        return 0.0

    # count occurrences of the token
    count = 0
    for w in words:
      count += 1 if w == token else 0

    # term frequency
    return count / len(words)
```

```python
def computeDF(token: str, corpus: list):
  # normalize token
  token = token.lower()

  # count docs with token
  count = 0
  for doc in corpus:

    # normalize doc and parse into words
    doc = doc.lower()
    words = re.findall(r"[a-z]+", doc)

    count += 1 if token in words else 0

  return count
```

```python
import math

def computeIDF(token: str, corpus: list):
  # calc values
  corpus_size = len(corpus)
  df = computeDF(token, corpus)

  return math.log(corpus_size / (df + 1))
```

```python
def computeTFIDF(token: str, doc: str, corpus: list):
  return computeTF(token, doc) * computeIDF(token, corpus)
```

```python
def sparse_search(query: str, corpus: list, num_results_to_return: int=3):
  # parse query
  query = query.lower()
  query_words = re.findall(r"[a-z]+", query)

  # compute TF-IDF scores for each doc
  scores = []
  for i, doc in enumerate(corpus):
    score = 0.0

    # compute TF-IDF in regards to a single token
    for token in query_words:
      score += computeTFIDF(token, doc, corpus)

    # catalog doc score
    scores.append((score, i))

  # sort docs by score
  sorted_scores = sorted(scores, key=lambda x: x[0], reverse=True)

  # return top results
  return [(corpus[i], score) for score, i in sorted_scores[:num_results_to_return]]

```

# Dense Search

### Implement

```python
from sentence_transformers import SentenceTransformer
import numpy as np
model_name = 'sentence-transformers/all-mpnet-base-v2'
```

```python
# Ingestion - run this only once, not per query

model = SentenceTransformer("sentence-transformers/all-mpnet-base-v2")
embedded_corpus = model.encode(corpus, convert_to_numpy=True)
print(embedded_corpus.shape)
```

```python
def dense_search(query: str, corpus: list, embedded_corpus: np.ndarray, model: SentenceTransformer, num_results_to_return: int=3):
  # embed the query
  q = model.encode(query, convert_to_numpy=True)

  # inner product of curpus and q
  scores = embedded_corpus @ q

  # get indices of top k docs
  top_k_indices = np.argsort(scores)[::-1][:num_results_to_return]

  # return the corresponding documents
  return [(corpus[i], float(scores[i])) for i in top_k_indices]
```
